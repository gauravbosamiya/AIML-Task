```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
```

```python
data1 = pd.read_csv("image_code_output.csv")
data1.head()
```

```python
data2 = pd.read_csv("AI ML Task Sheet.csv")
data2 = data2.loc[:, ~data2.columns.str.startswith('Unnamed')]
data2
```

```python
merged_data = pd.merge(data1, data2, on='code', how='inner')
merged_data_cleaned = merged_data.dropna()
```

```python
merged_data_cleaned['code'] = merged_data_cleaned['code'].astype('Int64')
merged_data_cleaned['qty'] = merged_data_cleaned['qty'].astype('Int64')
```

```python
df = merged_data_cleaned.copy()
df = df.groupby('code').agg({
    'image_path': 'first',
    'qty': 'sum',
    'rate': 'first',
    'date': 'first'
}).reset_index()
```

```python
df
```

```python
df.info()
```

```python
df['date'] = pd.to_datetime(df['date'])
```

```python
df['month'] = df['date'].dt.month
df['year'] = df['date'].dt.year
df
```

```python
import calendar
top3 = (
    df.groupby(['year', 'month', 'code'])['qty']
    .sum()
    .reset_index()
    .sort_values(['year', 'month', 'qty'], ascending=[True, True, False])
    .groupby(['year', 'month'])
    .head(3)
    .reset_index(drop=True)  
)

top3['month_name'] = top3['month'].apply(lambda x: calendar.month_abbr[x])
top3['period']     = top3['month_name'] + ' ' + top3['year'].astype(str)

top3[['period', 'code', 'qty']]
```

```python
most_sold = (
    df.groupby(['year', 'month', 'code'])['qty']
    .sum()
    .reset_index()
    .sort_values(['year', 'month', 'qty'], ascending=[True, True, False])
    .groupby(['year', 'month'])
    .first()
    .reset_index()
)

most_sold['month_name'] = most_sold['month'].apply(lambda x: calendar.month_abbr[x])
most_sold['period'] = most_sold['month_name'] + ' ' + most_sold['year'].astype(str)

print(most_sold[['period', 'code', 'qty']])
```
